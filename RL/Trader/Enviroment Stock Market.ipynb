```python
import numpy as np
import gym
import random

class Stock_Market(gym.Env):
    def __init__(self, data):
        """
        Inicializa a classe My_Env.

        Parâmetros:
        data (np.array): Array de estados representando o ambiente.
        """
        super().__init__()
        self.data = data
        self.observation_space = len(data)
        self.action_space = 3   # Exemplo: 0 - Manter, 1 - Comprar, 2 - Vender
        self.current_state = 0  # Índice do estado atual no ambiente
        self.done = False       # Indica se o episódio terminou
        self.inventory = []     # Inventário de ações compradas

    def reset(self, seed=None):
        """
        Reinicia o ambiente para o estado inicial.

        Retorna:
        estado_inicial (qualquer): O estado inicial do ambiente.
        """
        super().reset(seed=seed)
        self.current_state = 0
        self.done = False
        self.inventory = []  # Limpa o inventário ao reiniciar
        return self.data[self.current_state]

    def step(self, action):
        """
        Executa uma ação no ambiente.

        Parâmetros:
        action (int): A ação a ser tomada (0, 1 ou 2).

        Retorna:
        state (qualquer): O novo estado do ambiente.
        reward (float): A recompensa recebida pela ação tomada.
        done (bool): Indicador de término do episódio.
        """

        if self.done:
            raise Exception("O ambiente já terminou. Por favor, chame reset().")

        reward = 0
        price = self.data[self.current_state, 0]  # Preço da ação no estado atual

        if action == 1:  # Comprar ação
            self.inventory.append(price)

        elif action == 2 and len(self.inventory) > 0:    # Vender ação
            bought_price = self.inventory.pop(0)
            sell_price = self.data[self.current_state, 0]
            reward = sell_price - bought_price           # Lucro ou prejuízo da venda # Lucro ou prejuízo da venda

        # Avança para o próximo estado
        self.current_state += 1

        # Verifica se chegou ao fim do ambiente
        if self.current_state >= self.observation_space:
            self.done = True
            self.current_state = self.observation_space - 1  # Mantém no último estado

        state = self.data[self.current_state]
        return state, reward, self.done

    def sample(self):
      """
      Retorna uma ação aleatória.

      Retorna:
      action (int): Ação aleatória (0, 1 ou 2).
      """
      return random.choice(range(self.action_space))

    def __len__(self):
      """Retorna o tamanho atual da memória"""
      # Retorna o número de transições atualmente armazenadas na memória
      return self.observation_space

    def __getitem__(self, index):
     return self.data[index]
```
