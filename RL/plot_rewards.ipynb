```python
# plot_script.py

import matplotlib.pyplot as plt

def plot_rewards(rewards):
    """
    Plota os recompensas ao longo do tempo.

    Parameters:
    rewards (list): Uma lista de recompensas coletadas a cada episódio.
    """
    plt.figure(figsize=(10, 5))
    plt.plot(rewards)
    plt.xlabel('Episódios')
    plt.ylabel('Recompensa')
    plt.title('Recompensas ao longo do tempo')
    plt.show()

def plot_performance(metric, values):
    """
    Plota uma métrica de performance ao longo do tempo.

    Parameters:
    metric (str): Nome da métrica a ser plotada.
    values (list): Valores da métrica coletados a cada episódio.
    """
    plt.figure(figsize=(10, 5))
    plt.plot(values)
    plt.xlabel('Episódios')
    plt.ylabel(metric)
    plt.title(f'{metric} ao longo do tempo')
    plt.show()
```
