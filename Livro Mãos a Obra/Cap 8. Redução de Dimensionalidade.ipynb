```python
from sklearn.decomposition import  PCA
import numpy as np
```

```python
pca = PCA(n_components=2)
X = np.random.rand(200000).reshape(1000, 200)
X2D = pca.fit_transform(X)
```

```python
pca.explained_variance_ratio_
```

```python
pca = PCA()
pca.fit(X)
explained_variance = np.cumsum(pca.explained_variance_ratio_)
d = np.argmax(explained_variance >= 0.95) + 1
d
```

```python
pca = PCA(n_components=0.95)
X_reduced = pca.fit_transform(X)
X_reduced.shape
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 6))
plt.plot(range(1, len(explained_variance)+1), explained_variance, marker='o')
plt.xlabel('Número de Componentes Principais (Dimensões)')
plt.ylabel('Variância Explicada Acumulada')
plt.title('Gráfico do Cotovelo (Elbow Plot) - PCA')
plt.grid(True)

plt.axvline(x=20, color='r', linestyle='--', label='Elbow (~20)')
plt.legend()
plt.show()
```

```python
rnd_pca = PCA(n_components=154, svd_solver='randomized') # full, randomized, auto
X_reduced = rnd_pca.fit_transform(X)
```

```python
from sklearn.decomposition import IncrementalPCA

X = np.random.rand(200000).reshape(1000, 200)

n_batches = 100
inc_pca = IncrementalPCA(n_components=10)
for X_batch in np.array_split(X, n_batches):
    inc_pca.partial_fit(X_batch)

X_reduced = inc_pca.transform(X)
```

```python
from sklearn.decomposition import KernelPCA

rbf_pca = KernelPCA(n_components=2, kernel='rbf', gamma=0.04)
X_reduced = rbf_pca.fit_transform(X)
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline

X = np.random.rand(200000).reshape(1000, 200)
y = np.random.randint(0, 2, 1000)

clf = Pipeline([
    ('kpca', KernelPCA(n_components=2)),
    ('log_reg', LogisticRegression())
])

param_grid = [
    {
        "kpca__gamma": np.linspace(0.03, 0.05, 10),
        "kpca__kernel": ["rbf", "sigmoid"]
    }
]

grid_search = GridSearchCV(clf, param_grid, cv=3)
grid_search.fit(X, y)
```

```python
print(grid_search.best_params_)
```

```python
rbf_pca = KernelPCA(n_components=2, kernel='rbf', gamma=0.0433, fit_inverse_transform=True)
X_reduced = rbf_pca.fit_transform(X)
X_preimage = rbf_pca.inverse_transform(X_reduced)
```

```python
from sklearn.metrics import mean_squared_error
mean_squared_error(X, X_preimage)
```
