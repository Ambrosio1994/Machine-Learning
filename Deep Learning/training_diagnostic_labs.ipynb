```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
def load_training_data():
    """Load training logs for analysis"""
    # PROVIDED CODE - DO NOT MODIFY
    epochs = np.arange(1, 201)
    np.random.seed(42)

    # Simulate training curves with overfitting pattern
    train_loss = 2.5 * np.exp(-epochs/50) + 0.1 * np.random.normal(0, 0.1, 200)
    val_loss = 2.5 * np.exp(-epochs/60) + 0.3 * np.maximum(0, epochs-120)/50 + 0.15 * np.random.normal(0, 0.1, 200)

    # Simulate gradient norms with some instability
    grad_norms = 1.0 + 0.5 * np.random.normal(0, 0.2, 200)
    grad_norms[150:160] = grad_norms[150:160] * 5  # Simulate gradient spike

    return epochs, train_loss, val_loss, grad_norms
```

```python
def analyze_loss_curves(epochs, train_loss, val_loss):
    """Analyze training and validation loss patterns"""

    # Create a figure for plotting
    plt.figure(figsize=(10, 6))

    ### DESAFIO PRÁTICO 1 ###

    # Traçar curvas de perda
    plt.plot(epochs, train_loss, label="Training Loss")
    plt.plot(epochs, val_loss, label="Validation Loss")

    # Detectar início do overfitting
    overfitting_epoch = None

    for i in range(1, len(epochs)):
        # Overfitting:
        # training loss diminui enquanto validation loss aumenta
        if train_loss[i] < train_loss[i - 1] and val_loss[i] > val_loss[i - 1]:
            overfitting_epoch = epochs[i]

            # Marcar ponto no gráfico
            plt.scatter(
                epochs[i],
                val_loss[i],
                s=100,
                label=f"Overfitting starts (epoch {epochs[i]})"
            )

            # Linha vertical indicando a época
            plt.axvline(
                x=epochs[i],
                linestyle="--"
            )

            break

    # Mensagem no console
    if overfitting_epoch is not None:
        print(f"Overfitting detected starting at epoch {overfitting_epoch}.")
    else:
        print("No overfitting detected.")

    # Configuração do gráfico
    plt.xlabel("Epoch")
    plt.ylabel("Loss")
    plt.title("Training and Validation Loss")
    plt.legend()
    plt.grid(True)
    plt.show()

    return overfitting_epoch
```

```python
def analyze_gradient_norms(epochs, grad_norms):
    """Analyze gradient norm patterns for stability issues"""

    plt.figure(figsize=(10, 6))

    ### DESAFIO PRÁTICO 2 ###

    threshold = 3.0
    window_size = 3

    # Traçar normas de gradiente
    plt.plot(
        epochs,
        grad_norms,
        label="Gradient Norm"
    )

    # Calcular média móvel
    moving_avg = []

    for i in range(len(grad_norms)):
        start = max(0, i - window_size + 1)
        window = grad_norms[start:i + 1]
        moving_avg.append(sum(window) / len(window))

    # Traçar média móvel
    plt.plot(
        epochs,
        moving_avg,
        linestyle="--",
        label=f"Moving Average ({window_size} epochs)"
    )

    # Linha indicando limite considerado perigoso
    plt.axhline(
        y=threshold,
        linestyle="--",
        label=f"Threshold ({threshold})"
    )

    # Detectar todos os picos
    spike_epochs = []

    for i in range(len(epochs)):
        if grad_norms[i] > threshold:
            spike_epochs.append(epochs[i])

            plt.scatter(
                epochs[i],
                grad_norms[i],
                s=100
            )

    # Configuração do gráfico
    plt.xlabel("Epoch")
    plt.ylabel("Gradient Norm")
    plt.title("Gradient Norms During Training")
    plt.legend()
    plt.grid(True)

    plt.show()

    return bool(spike_epochs)
```

```python
def generate_diagnostic_report(overfitting_epoch, gradient_issues):
    """Generate comprehensive diagnostic report"""

    print("\n" + "=" * 60)
    print("NEURAL NETWORK TRAINING DIAGNOSTIC REPORT")
    print("=" * 60)

    # 1. Diagnóstico de overfitting
    print("\n1. OVERFITTING ANALYSIS")
    print("-" * 60)

    if overfitting_epoch is not None:
        print(f"Status: Overfitting detected")
        print(f"Starting epoch: {overfitting_epoch}")
        print(
            "Recommendation: consider using early stopping, "
            "regularization, dropout, or additional training data."
        )
    else:
        print("Status: No overfitting detected.")
        print(
            "Recommendation: current training/validation behavior "
            "does not indicate an obvious overfitting problem."
        )

    # 2. Diagnóstico dos gradientes
    print("\n2. GRADIENT STABILITY ANALYSIS")
    print("-" * 60)

    if gradient_issues:
        print("Status: Gradient instability detected.")
        print(
            "Recommendation: consider applying gradient clipping "
            "to limit excessive gradient norms."
        )
        print(
            "Also consider reducing the learning rate if "
            "gradient spikes continue."
        )
    else:
        print("Status: Gradients appear stable.")
        print(
            "Recommendation: no gradient-specific intervention "
            "is currently required."
        )

    # 3. Recomendações para o próximo treino
    print("\n3. RECOMMENDATIONS FOR NEXT TRAINING RUN")
    print("-" * 60)

    if overfitting_epoch is not None:
        print(
            f"- Configure early stopping around epoch "
            f"{overfitting_epoch}."
        )
        print("- Consider increasing regularization.")
        print("- Consider adding or augmenting training data.")

    if gradient_issues:
        print("- Enable gradient clipping.")
        print("- Monitor gradient norms during training.")
        print("- Consider lowering the learning rate.")

    if overfitting_epoch is None and not gradient_issues:
        print("- No major training issues detected.")
        print("- Continue monitoring validation loss and gradient norms.")

    print("\n" + "=" * 60)
    print("END OF DIAGNOSTIC REPORT")
    print("=" * 60)
```

```python
print("Loading training data...")
epochs, train_loss, val_loss, grad_norms = load_training_data()

print("Analyzing loss curves...")
overfitting_epoch = analyze_loss_curves(epochs, train_loss, val_loss)

print("Analyzing gradient norms...")
gradient_issues = analyze_gradient_norms(epochs, grad_norms)
```

```python
print("Generating diagnostic report...")
generate_diagnostic_report(overfitting_epoch, gradient_issues)

print("Analysis complete!")
```
