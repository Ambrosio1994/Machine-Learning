````python
import kagglehub
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
````

````python
# Download latest version
path = kagglehub.dataset_download("yusufdelikkaya/datascience-salaries-2024")

print("Path to dataset files:", path)
print()
print(os.listdir(path))
````

````python
path_csv = path + "/DataScience_salaries_2024.csv"
pd.options.display.float_format = '{:.2f}'.format
df = pd.read_csv(path_csv)
df.head()
````

````python
df = df.groupby(['work_year', 'experience_level','job_title','company_size', 'company_location'])['salary_in_usd'].mean().reset_index()
df
````

````python
print(df["experience_level"].unique())
````

````python
mid_level_df = df[df["experience_level"] == "MI"]
senior_df = df[df["experience_level"] == "SE"]
entry_level_df = df[df["experience_level"] == "EN"]
executive_df = df[df["experience_level"] == "EX"]
````

````python
entry_level_df.head()
````

````python
plt.figure(figsize=(12, 6))
sns.barplot(x=entry_level_df["company_location"], y=entry_level_df["salary_in_usd"])
plt.xticks(rotation=90)
plt.xlabel("Company Location")
plt.ylabel("Salary in USD")
plt.title("Top 10 Entry-Level Jobs by Company Location")
plt.tight_layout()
plt.show()
````
