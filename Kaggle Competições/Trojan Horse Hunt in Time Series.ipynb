```python
def load_poised_models(path_poisened_models):
  poisoned_model_paths  = [f"{path_poisened_models}/poisoned_model_{model_id}/poisoned_model.pt" for model_id in tqdm(range(1, 46))]
  if len(poisoned_model_paths) == 45:
    print("✅ Carregamento concluído")
  poisoned_models = [NHiTSModel.load(path) for path in poisoned_model_paths]
  if len(poisoned_models) == 45:
    print("✅ Modelos criados")
  return poisoned_models
```
