```python
import cv2
from google.colab import drive
drive.mount('/content/drive')
```

```python
imagem = cv2.imread('/content/drive/MyDrive/Images/people1.jpg')
imagem.shape
```

```python
from google.colab.patches import cv2_imshow
cv2_imshow(imagem)
```

```python
imagem = cv2.resize(imagem,(800, 600))
cv2_imshow(imagem)
```

```python
imagem_cinza = cv2.cvtColor(imagem, cv2.COLOR_BGR2GRAY)
cv2_imshow(imagem_cinza)
```

```python
detector_facil = cv2.CascadeClassifier('/content/drive/MyDrive/Visão Computacional Guia Completo/Cascades/haarcascade_frontalface_default.xml')
```

```python
deteccoes = detector_facil.detectMultiScale(imagem_cinza, scaleFactor=1.09)
deteccoes
```

```python
imagem2 = cv2.imread('/content/drive/MyDrive/Visão Computacional Guia Completo/Images/people2.jpg')
imagem_cinza2 = cv2.cvtColor(imagem2, cv2.COLOR_BGR2GRAY)
deteccoes2 = detector_facil.detectMultiScale(imagem_cinza2, scaleFactor=1.02,
                                            minNeighbors=20, minSize=(30,30),
                                            maxSize=(100, 100))
for x, y, w, h in deteccoes2:
  cv2.rectangle(imagem2, (x, y), (x + w, y + h), (0, 255, 0), 5)

cv2_imshow(imagem2)
```

```python
detector_olhos = cv2.CascadeClassifier('/content/drive/MyDrive/Visão Computacional Guia Completo/Cascades/haarcascade_eye.xml'
)
```

```python
imagem = cv2.imread('/content/drive/MyDrive/Images/people1.jpg')

imagem_cinza = cv2.cvtColor(imagem, cv2.COLOR_BGR2GRAY)

deteccoes_faciais = detector_facil.detectMultiScale(imagem_cinza, scaleFactor=1.3,
                                            minSize=(30,30)
                                            )
for (x, y, w, h) in deteccoes_faciais:
  cv2.rectangle(imagem, (x, y), (x + w, y + h), (0, 255, 0), 5)


deteccoes_olhos = detector_olhos.detectMultiScale(imagem_cinza, scaleFactor=1.09,
                                            minNeighbors=10,
                                            maxSize=(70, 70)
                                            )
for (x, y, w, h) in deteccoes_olhos:
  cv2.rectangle(imagem, (x, y), (x + w, y + h), (0, 0, 255), 2)
cv2_imshow(imagem)
```
