```python
import tensorflow as tf
import numpy as np
from tensorflow.keras.datasets import imdb
```

```python
tf.__version__
```

```python
NUMBER_OF_WORDS = 20000
MAX_LEN = 100
```

```python
(X_train, y_train), (X_test, y_test) = imdb.load_data(num_words=NUMBER_OF_WORDS)
```

```python
# 1 positivo, 0 negativo
```

```python
X_train = tf.keras.preprocessing.sequence.pad_sequences(X_train, maxlen=MAX_LEN)
X_test = tf.keras.preprocessing.sequence.pad_sequences(X_test, maxlen=MAX_LEN)
```

```python
len(X_train[0]), len(X_test[0])
```

```python
model = tf.keras.Sequential([
    tf.keras.layers.Embedding(input_dim=NUMBER_OF_WORDS, output_dim=128, input_shape=(X_train.shape[1],)),
    tf.keras.layers.LSTM(units=128, activation="tanh"),
    tf.keras.layers.Dense(units=1, activation="sigmoid")
])
```

```python
model.compile(
    optimizer="rmsprop", loss="binary_crossentropy", metrics="accuracy")
```

```python
model.fit(
    X_train, y_train,epochs=3, batch_size=128, validation_data=(X_test, y_test))
```
