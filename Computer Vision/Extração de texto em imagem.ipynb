```python
import cv2
import numpy as np
from pdf2image import convert_from_path
from pytesseract import image_to_string
```

```python
pdf_path = '/content/DHD52407093.pdf'
```

```python
def extract_code(pdf_path):
    # Converta a primeira página do PDF em imagens
    images = convert_from_path(pdf_path, first_page=1, last_page=1)

    if images:
        # Converta a primeira imagem para um formato adequado
        image = images[0].convert('RGB')
        img = np.array(image)

        # Extraia a região especificada da imagem
        barcode_region = [1098, 129, 165, 28]
        x, y, w, h = barcode_region
        extracted_region = img[y:y + h, x:x + w]

        # Converta a região extraída para texto usando OCR
        extracted_text = image_to_string(extracted_region)
        text = extracted_text.split()[0]

        return text
```

```python
text = extract_code(pdf_path)
```
