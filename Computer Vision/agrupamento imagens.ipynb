```python
import dlib
import cv2
from google.colab.patches import cv2_imshow
import os
```

```python
def criar_pasta(pasta):
    if not os.path.exists(pasta):
        os.makedirs(pasta)
```

```python
diretorio_imagens = "/content/drive/MyDrive/Agrupamento"
```

```python
diretorio_base = "Pessoas"
```

```python
detector = dlib.get_frontal_face_detector()
```

```python
imagens = [imagem for imagem in os.listdir(diretorio_imagens) if imagem.lower().endswith(('.png', '.jpg', '.jpeg', '.gif'))]
```

```python
for imagem_nome in imagens:
    # Carregando a imagem
    caminho_imagem = os.path.join(diretorio_imagens, imagem_nome)
    imagem = cv2.imread(caminho_imagem)
    imagem_cinza = cv2.cvtColor(imagem, cv2.COLOR_BGR2GRAY)

    faces = detector(imagem_cinza, 2)

    for i, face in enumerate(faces):
        x, y, w, h = face.left(), face.top(), face.width(), face.height()

        # Criando uma pasta para cada pessoa
        pasta_pessoa = os.path.join(diretorio_base, f"Pessoa_{i + 1}")
        criar_pasta(pasta_pessoa)

        # Salvando a imagem da pessoa na pasta correspondente
        caminho_imagem_pessoa = os.path.join(pasta_pessoa, f"{imagem_nome[:-4]}_rosto_{i + 1}.jpg")
        cv2.imwrite(caminho_imagem_pessoa, imagem[y:y+h, x:x+w])

print("Processo concluído. Verifique as subpastas em", diretorio_base)
```

```python
imagem = cv2.imread('/content/drive/MyDrive/Agrupamento/DSC02137-288.jpg')
imagem_cinza = cv2.cvtColor(imagem, cv2.COLOR_BGR2GRAY)
faces = detector(imagem_cinza, 2)

numero_pessoa = 1

for face in faces:
    x, y, w, h = face.left(), face.top(), face.width(), face.height()

    # Criando uma pasta para cada pessoa
    pasta_pessoa = os.path.join(diretorio_base, f"Pessoa_{numero_pessoa}")
    criar_pasta(pasta_pessoa)

    # Salvando a imagem da pessoa na pasta correspondente
    caminho_imagem_pessoa = os.path.join(pasta_pessoa, f"imagem_{numero_pessoa}.jpg")
    cv2.imwrite(caminho_imagem_pessoa, imagem[y:y+h, x:x+w])

    # Adicionando anotações na imagem original
    cv2.rectangle(imagem, (x, y), (x + w, y + h), (0, 255, 0), 2)
    cv2.putText(imagem, f"Pessoa {numero_pessoa}", (x, y - 10), cv2.FONT_HERSHEY_SIMPLEX, 0.9, (0, 255, 0), 2)

    numero_pessoa += 1
cv2_imshow(imagem)
```
