```python
import cv2
import matplotlib.pyplot as plt
```

```python
def show_detection(path):
  img = cv2.imread(path)
  figure = plt.gcf()
  figure.set_size_inches(18, 10)
  plt.axis('off')
  plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
```

```python
show_detection('/content/darknet/predictions.jpg')
```
