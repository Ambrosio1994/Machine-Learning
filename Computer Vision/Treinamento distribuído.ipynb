```python
import time
import numpy as np
import tensorflow as tf
```

```python
(X_train, y_train), (X_test, y_test) = tf.keras.datasets.mnist.load_data()
```

```python
X_train = X_train / 255.0
X_test = X_test / 255.0
```

```python
X_train = X_train.reshape(-1, 28*28)
X_test = X_test.reshape(-1, 28*28)
```

```python
model1 = tf.keras.models.Sequential()
model1.add(tf.keras.layers.Dense(128, activation="relu", input_shape=(784,)))
model1.add(tf.keras.layers.Dropout(0.2))
model1.add(tf.keras.layers.Dense(10))
model1.compile(optimizer="adam", loss="sparse_categorical_crossentropy", metrics=["sparse_categorical_accuracy"])
```

```python
distribute = tf.distribute.MirroredStrategy()
```

```python
with distribute.scope():
  model2 = tf.keras.models.Sequential()
  model2.add(tf.keras.layers.Dense(128, activation="relu", input_shape=(784,)))
  model2.add(tf.keras.layers.Dropout(0.2))
  model2.add(tf.keras.layers.Dense(10))
  model2.compile(optimizer="adam", loss="sparse_categorical_crossentropy", metrics=["sparse_categorical_accuracy"])
```

```python
starting_time = time.time()
model2.fit(X_train, y_train, epochs=10, batch_size=128, validation_data=(X_test, y_test))
print(f'Treinamento distribuído: {time.time() - starting_time}')
```
