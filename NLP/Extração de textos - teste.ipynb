```python
from difflib import SequenceMatcher
from pytesseract import image_to_string
from google.colab.patches import cv2_imshow
import re
import cv2
import pytesseract
from pytesseract import Output
```

```python
imagem1 = '/content/estepyton.jpg'
imagem2 = '/content/supermercados.png'
imagem3 = '/content/drive/MyDrive/Images/trecho-livro.jpg'
```

```python
def extrair_texto(imagem, lang='por', config='--tessdata-dir tessdata --psm 3'):
    try:
        img = cv2.imread(imagem)
        img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

        # Preprocess the image for better OCR accuracy:
        gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
        if gray.mean() < 127:  # Adaptive thresholding for better contrast
            thresh = cv2.threshold(gray, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)[1]
        else:
            thresh = cv2.threshold(gray, 127, 255, cv2.THRESH_BINARY)[1]

        # Apply noise reduction if needed:
        thresh = cv2.medianBlur(thresh, 1)  # Uncomment for median filtering

        # Extract text using Tesseract:
        d = pytesseract.image_to_string(thresh, lang=lang, config=config, output_type=Output.DICT)

        # Clean the extracted text:
        texto = d['text']
        texto = re.sub(r'\W+', ' ', texto)  # Remove all non-alphanumeric characters

        return texto

    except Exception as e:
        print(f"Erro ao extrair texto da imagem: {e}")
        return None
```

```python
texto1 = extrair_texto(imagem1)
texto2 = extrair_texto(imagem2)
texto3 = extrair_texto(imagem3)
```

```python
# Função para comparar palavras
def comparar_palavras(palavra1, palavra2):
    # Use SequenceMatcher para comparar palavras
    similaridade = SequenceMatcher(None, palavra1, palavra2).ratio()
    return similaridade
```

```python
# Comparar palavras
similaridade = comparar_palavras(texto1, texto2)
print(f"Similaridade entre textos: {similaridade}")
```

```python
import tensorflow as tf
from tensorflow.keras import datasets, layers, models
```

```python
# Carregar e dividir os dados
(train_images, train_labels), (test_images, test_labels) = datasets.cifar10.load_data()

# Normalizar os valores dos pixels para estarem entre 0 e 1
train_images, test_images = train_images / 255.0, test_images / 255.0
```

```python
# Definir o modelo
model = models.Sequential()
model.add(layers.Conv2D(32, (3, 3), activation='relu', input_shape=(32, 32, 3)))
model.add(layers.MaxPooling2D((2, 2)))
model.add(layers.Conv2D(64, (3, 3), activation='relu'))
model.add(layers.MaxPooling2D((2, 2)))
model.add(layers.Conv2D(64, (3, 3), activation='relu'))

# Adicionar mais camadas
model.add(layers.Flatten())
model.add(layers.Dense(64, activation='relu'))
model.add(layers.Dense(10))
```

```python
# Compilar e treinar o modelo
model.compile(optimizer='adam',
              loss=tf.keras.losses.SparseCategoricalCrossentropy(from_logits=True),
              metrics=['accuracy'])

history = model.fit(train_images, train_labels, epochs=10,
                    validation_data=(test_images, test_labels))
```

```python
# Importar as bibliotecas necessárias
import torch
from detectron2 import model_zoo
from detectron2.config import get_cfg
from detectron2.engine import DefaultPredictor
from detectron2.data import MetadataCatalog
from detectron2.utils.visualizer import Visualizer
from PIL import Image
import cv2

# Configuração do modelo
cfg = get_cfg()
cfg.merge_from_file(model_zoo.get_config_file("COCO-Detection/faster_rcnn_R_50_FPN_3x.yaml"))
cfg.MODEL.ROI_HEADS.SCORE_THRESH_TEST = 0.5  # Define um limiar de confiança para as previsões

# Carregar o modelo pré-treinado
cfg.MODEL.WEIGHTS = model_zoo.get_checkpoint_url("COCO-Detection/faster_rcnn_R_50_FPN_3x.yaml")

# Criar o preditor
predictor = DefaultPredictor(cfg)

# Carregar uma imagem de exemplo (você pode substituir isso pelo caminho da sua própria imagem)
image = cv2.imread('/content/supermercados.png')

# Carregar a imagem diretamente do caminho do arquivo
#image = Image.open(image_path)

# Fazer a previsão na imagem
outputs = predictor(image)

# Visualizar os resultados
v = Visualizer(image, MetadataCatalog.get(cfg.DATASETS.TRAIN[0]), scale=1.2)
out = v.draw_instance_predictions(outputs["instances"].to("cpu"))
Image.fromarray(out.get_image())
```
