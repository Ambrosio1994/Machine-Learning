```python
def extract_data(caminho):
  data = pd.read_excel(caminho, sheet_name=["Início", "Síntese das emissões",  "Formulário de emissões"])

  dados_inicio = data["Início"]
  dados_sintese = data["Síntese das emissões"]
  dados_formulario = data["Formulário de emissões"]

  nome_org = dados_inicio.loc[11, 'Unnamed: 2']
  endereco = dados_inicio.loc[12, 'Unnamed: 2']
  nome_resp = dados_inicio.loc[14, 'Unnamed: 2']
  periodo = dados_inicio.loc[17, 'Unnamed: 2']

  estado = dados_formulario.loc[26, 'Unnamed: 1']
  bioma = dados_formulario.loc[26, 'Unnamed: 2']
  area_total = dados_formulario.loc[26, 'Unnamed: 4']
  area_pastagem = dados_formulario.loc[83, 'Unnamed: 3']

  rebanho = ''
  escopos_processo_mudanca = dados_sintese.loc[32, 'Unnamed: 4']
  escopo_1 = dados_sintese.loc[16, 'Unnamed: 8']
  escopo_2 = dados_sintese.loc[17, 'Unnamed: 8']
  escopo_3 = dados_sintese.loc[18, 'Unnamed: 8']
  uso_mudanca_solo = dados_sintese.loc[22, 'Unnamed: 4']
  emissao_biogenica = dados_sintese.loc[26, 'Unnamed: 4']
  remocao_biogenica = dados_sintese.loc[28, 'Unnamed: 4']

  extract = {
        'nome_org': nome_org,
        'endereco': endereco,
        'nome_resp': nome_resp,
        'periodo': periodo,
        'estado': estado,
        'bioma': bioma,
        'area_total': area_total,
        'area_pastagem': area_pastagem,
        'rebanho': rebanho,
        'escopos_processo_mudanca': escopos_processo_mudanca,
        'escopo_1': escopo_1,
        'escopo_2': escopo_2,
        'escopo_3': escopo_3,
        'uso_mudanca_solo': uso_mudanca_solo,
        'emissao_biogenica': emissao_biogenica,
        'remocao_biogenica': remocao_biogenica
    }
  return extract
```

```python
def insert(df, dictionary):
  """Insere as informações de um dicionário em um DataFrame.

  Args:
    df: Um objeto `DataFrame` do Pandas.
    dictionary: Um dicionário com as informações a serem inseridas.

  Returns:
    O DataFrame com as informações inseridas.
  """

  # Cria uma lista com as chaves do dicionário.
  keys = list(dictionary.keys())

  # Cria uma lista com os valores do dicionário.
  values = list(dictionary.values())

  # Insere as informações na última linha do DataFrame.
  df.loc[len(df)] = values

  # Define as chaves como nomes das colunas.
  #df.columns = keys

  return df
```
