```python
# Importando bibliotecas necessárias
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
# Dados de exemplo
base_de_documentos = [
    "Exemplo de documento sobre machine learning.",
    "Introdução à análise de dados usando Python.",
    "Como aplicar regressão linear em ciência de dados.",
    # Adicione mais documentos conforme necessário
]
```

```python
# Função para buscar textos com base no assunto fornecido pelo usuário
def buscar_textos(assunto, num_exemplos=3):
    # Adicionando o assunto do usuário aos documentos de exemplo
    base_de_documentos.append(assunto)

    # Convertendo os documentos para vetores usando TF-IDF
    tfidf_vectorizer = TfidfVectorizer()
    tfidf_matrix = tfidf_vectorizer.fit_transform(base_de_documentos)

    # Calculando a similaridade de cosseno entre o assunto do usuário e os documentos existentes
    similarity_scores = cosine_similarity(tfidf_matrix[-1], tfidf_matrix[:-1])

    # Obtendo os índices dos documentos mais semelhantes
    top_indices = similarity_scores.argsort()[0][-num_exemplos:][::-1]

    # Exibindo os textos mais relevantes para o usuário
    textos_encontrados = [base_de_documentos[i] for i in top_indices]
    return textos_encontrados
```

```python
# Solicitando entrada do usuário

assunto_usuario = input("Digite o assunto que você quer buscar: ")
num_exemplos_usuario = int(input("Quantos textos de exemplo você deseja encontrar? "))

# Obtendo resultados e exibindo para o usuário
resultados = buscar_textos(assunto_usuario, num_exemplos_usuario)
print("\nTextos encontrados:")
for i, texto in enumerate(resultados, 1):
    print(f"{i}. {texto}")
```
