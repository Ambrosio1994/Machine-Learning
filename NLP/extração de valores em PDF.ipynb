```python
import PyPDF2

def extrair_notas_servico(arquivo_pdf, config=1):
    # Abre o arquivo PDF
    with open(arquivo_pdf, "rb") as pdf_file:
        # Cria um objeto PdfReader
        pdf_reader = PyPDF2.PdfReader(pdf_file)

        # Obtém o número de páginas do PDF
        num_pages = len(pdf_reader.pages)

        # Configurações
        if config == 1:
            campos = [53, 1, 3, 61, 63]
        elif config == 2:
            campos = [9, 7, 16, 49, 44]
        elif config == 3:
            campos = [16, 12, 27, 33, 44]
        else:
            raise ValueError("Configuração inválida. Escolha 1, 2 ou 3.")

        # Itera pelas páginas do PDF
        for page_num in range(num_pages):
            # Obtém a página atual
            page = pdf_reader.pages[page_num]

            # Extrai as informações da nota de serviço (usando extract_text())
            texto_pagina = page.extract_text()

            # Coleta informações conforme a configuração
            data_emissao = texto_pagina.split("\n")[campos[0]]
            numero_nota = texto_pagina.split("\n")[campos[1]]
            cnpj_prestador = texto_pagina.split("\n")[campos[2]]
            descricao_servicos = texto_pagina.split("\n")[campos[3]]
            valor_total = texto_pagina.split("\n")[campos[4]]

            # Cria uma nota de serviço
            nota_servico = {
                "Data emissao": data_emissao,
                "Número da nota": numero_nota,
                "CNPJ do prestador": cnpj_prestador,
                "Descricao do servicos": descricao_servicos,
                "Valor total": valor_total,
            }

            # Retorna a nota de serviço
            return nota_servico
```

```python
# Importar os módulos PyPDF2 e re
import PyPDF2
import re

# Definir as constantes dos campos
DATA_EMISSAO = 53
NUMERO_NOTA = 1
CNPJ_PRESTADOR = 3
DESCRICAO_SERVICOS = 61
VALOR_TOTAL = 63

# Definir o dicionário das configurações
configuracoes = {
    1: [DATA_EMISSAO, NUMERO_NOTA, CNPJ_PRESTADOR, DESCRICAO_SERVICOS, VALOR_TOTAL],
    2: [9, 7, 16, 49, 44],
    3: [16, 12, 27, 33, 44],
}

# Definir a função que extrai as notas de serviço de um arquivo PDF
def extrair_notas_servico(arquivo_pdf, config=1):
    # Verificar se a configuração é válida
    if config not in configuracoes:
        raise ValueError("Configuração inválida. Escolha 1, 2 ou 3.")

    # Tentar abrir e ler o arquivo PDF
    try:
        with open(arquivo_pdf, "rb") as arquivo:
            # Criar um objeto PdfReader
            leitor_pdf = PyPDF2.PdfReader(arquivo)

            # Obter o número de páginas do PDF
            num_paginas = len(leitor_pdf.pages)

            # Iterar pelas páginas do PDF
            for num_pagina in range(num_paginas):
                # Obter a página atual
                pagina = leitor_pdf.pages[num_pagina]

                # Extrair as informações da nota de serviço
                texto_pagina = pagina.extract_text()
                informacoes = [texto_pagina.split("\n")[campo] for campo in configuracoes[config]]

                # Substituir os caracteres que não sejam dígitos por uma string vazia nos campos desejados
                informacoes[1] = re.sub("\D", "", informacoes[1]) # Número da nota
                informacoes[2] = re.sub("\D", "", informacoes[2]) # CNPJ do prestador
                #informacoes[4] = re.sub("\D", "", informacoes[4]) # Valor total

                # Criar uma nota de serviço
                nota_servico = dict(zip(["Data emissao", "Número da nota", "CNPJ do prestador", "Descricao do servicos", "Valor total"], informacoes))
                # Retornar a nota de serviço
                return nota_servico

    except PyPDF2.PdfReadError:
        print(f"O arquivo {arquivo_pdf} não é um PDF válido.")
```
